This notebook will help us to check the count of incremental data present in some foreign table, based on the count set in the parameter row_count using taskvalues, I am going to decide whether to  run the subsequent ETL tasks/pipeline or not.

```python
%sql
SELECT *
    FROM gcp_mysql_fc_wd36.logistics.shipments1
```

```python
row_count = spark.sql("""
    SELECT *
    FROM gcp_mysql_fc_wd36.logistics.shipments1
    WHERE updated_at > (
        SELECT COALESCE(MAX(updated_at), '1970-01-01')
        FROM catalog3_we47.schema3_we47.bronze_shipments1
    )
""").count()

print(f"Found {row_count:,} new rows in the MySQL table.")
```

```python
#this is a job task level property to set a parameter/variable inside a task, which can be propagated/read by other task on the same lakeflow job
dbutils.jobs.taskValues.set(key="row_count", value=row_count)
```
